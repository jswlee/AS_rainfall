```python
# Open daily features NPZ file
import numpy as np

npz_path = Path(project_dir / "output/daily_features.npz")
data = np.load(npz_path, allow_pickle=True)

# Access the 'variables' array
variables = data['variables']
print("Variables:", variables)
print("-" * 20)

# Access the 'meta' array (station and date info)
meta = data['meta']
print("Shape of meta array:", meta.shape)
print("First meta entry:", meta[0])
print("-" * 20)

# Access the 'patches' array
patches = data['patches']
print("Shape of patches array:", patches.shape)
print("(Samples, Variables, Height, Width)")
print("-" * 20)

# Print the patch for the first sample, first variable
print("First patch (for variable '{}' at '{}' for station '{}'):".format(
    variables[0], meta[0]['date'], meta[0]['station']
))
print(patches[0, 0, :, :])
```
