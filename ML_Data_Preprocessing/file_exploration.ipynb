```python
import xarray as xr
from pathlib import Path
import numpy as np

data_dir = Path("raw_data/AS/climate_variables_daily_FULL_1980-2024")

print("Data directory:", data_dir)
print("Exists:", data_dir.exists())

nc_files = sorted(data_dir.glob("*.nc"))
print("Number of .nc files found:", len(nc_files))
print("Files:", [p.name for p in nc_files])
print()

if not nc_files:
    print("No NetCDF files found – check the path above.")
else:
    target_start = np.datetime64("1980-01-01")
    target_end   = np.datetime64("2024-12-31")

    for nc_path in nc_files:
        try:
            with xr.open_dataset(nc_path) as ds:
                # Try to find a time-like coordinate
                if "valid_time" in ds.coords:
                    tcoord_name = "valid_time"
                elif "time" in ds.coords:
                    tcoord_name = "time"
                else:
                    # fallback: any coord with datetime dtype
                    time_like = [
                        name for name, coord in ds.coords.items()
                        if np.issubdtype(coord.dtype, np.datetime64)
                    ]
                    if time_like:
                        tcoord_name = time_like[0]
                    else:
                        print(f"{nc_path.name}: no time-like coordinate found")
                        continue

                t = ds[tcoord_name].values
                if t.size == 0:
                    print(f"{nc_path.name}: empty {tcoord_name} coordinate")
                    continue

                t_min = t.min()
                t_max = t.max()

                overlaps = (t_max >= target_start) and (t_min <= target_end)

                print(f"{nc_path.name}:")
                print(f"  time coord: {tcoord_name}")
                print(f"  available from {np.datetime_as_string(t_min)} "
                      f"to {np.datetime_as_string(t_max)}")
                print(f"  overlaps 1980–2024: {overlaps}")
                print("  variables:", ", ".join(sorted(ds.data_vars)))
                print()
        except Exception as e:
            print(f"{nc_path.name}: ERROR -> {e}")
```
